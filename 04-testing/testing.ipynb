```python
count = setup_database()
```

```python
tools = SQLTools(con)

print(tools.get_schema())

print(
    tools.run_sql("""
        SELECT
            passenger_count,
            AVG(total_amount) AS avg_fare
        FROM trips
        GROUP BY passenger_count
        ORDER BY avg_fare DESC
    """)
)
```

```python
config = SQLAgentConfig()
tools = SQLTools(con)
agent = create_agent(config, tools)
```

```python
result = await run_agent(
    agent,
    "What's the average trip distance for rides with 2 passengers?"
)
```
